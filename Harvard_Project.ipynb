```python
object_url = "https://api.harvardartmuseums.org/object"

classification="Prints"
all_records = []

page = 1

while len(all_records) < 2500:
    
    params = {
        "apikey": api_key,
        "classification": classification,
        "size": 100,
        "page": page
    }
    
    response = requests.get(object_url,params)
    data = response.json()
    
    records = data["records"]
    
    all_records.extend(records)
    
    print("Page:", page, "Records collected:", len(all_records))
    
    if len(records) == 0:
        break
    
    page += 1

print("Finished!")
print("Total records:", len(all_records))    
```

```python
all_artifacts = []

all_artifacts.extend(all_records)
all_artifacts.extend(all_paintings)
all_artifacts.extend(all_sculpture)
all_artifacts.extend(all_coins)
all_artifacts.extend(all_calligraphy)

print("Total artifacts:", len(all_artifacts))
```

```python
metadata_data = []
media_data = []
colors_data = []

for record in all_artifacts:

    metadata_data.append((
        record.get("id"),
        record.get("title"),
        record.get("culture"),
        record.get("period"),
        record.get("century"),
        record.get("medium"),
        record.get("dimensions"),
        record.get("description"),
        record.get("department"),
        record.get("classification"),
        record.get("accessionyear"),
        record.get("accessionmethod")
    ))

    media_data.append((
        record.get("id"),
        record.get("imagecount"),
        record.get("mediacount"),
        record.get("colorcount"),
        record.get("rank"),
        record.get("datebegin"),
        record.get("dateend")
    ))

    if record.get("colors"):
        for color in record["colors"]:
            colors_data.append((
                record.get("id"),
                color.get("color"),
                color.get("spectrum"),
                color.get("hue"),
                color.get("percent"),
                color.get("css3")
            ))

print("Metadata records:", len(metadata_data))
print("Media records:", len(media_data))
print("Color records:", len(colors_data))
```

```python
create_metadata_table = """
CREATE TABLE artifact_metadata (
    id INT PRIMARY KEY,
    title NVARCHAR(500),
    culture NVARCHAR(255),
    period NVARCHAR(255),
    century NVARCHAR(100),
    medium NVARCHAR(MAX),
    dimensions NVARCHAR(MAX),
    description NVARCHAR(MAX),
    department NVARCHAR(255),
    classification NVARCHAR(255),
    accessionyear INT,
    accessionmethod NVARCHAR(255)
)
"""

cursor.execute(create_metadata_table)
connection.commit()

print("Metadata table created successfully!")
```

```python
create_media_table = """
CREATE TABLE artifact_media (
    objectid INT PRIMARY KEY,
    imagecount INT,
    mediacount INT,
    colorcount INT,
    [rank] INT,
    datebegin INT,
    dateend INT
)
"""

cursor.execute(create_media_table)
connection.commit()

print("Media table created successfully!")
```

```python
create_colors_table = """
CREATE TABLE artifact_colors (
    objectid INT,
    color NVARCHAR(MAX),
    spectrum NVARCHAR(MAX),
    hue NVARCHAR(MAX),
    [percent] FLOAT,
    css3 NVARCHAR(MAX)
)
"""

cursor.execute(create_colors_table)
connection.commit()

print("Colors table created successfully!")
```

```python
query20 = """
SELECT
    m.classification,
    COUNT(*) AS total_artifacts,
    AVG(a.mediacount) AS average_media_count
FROM artifact_metadata AS m
INNER JOIN artifact_media AS a
    ON m.id = a.objectid
GROUP BY m.classification
ORDER BY total_artifacts DESC;
"""
cursor.execute(query20)
results20 = cursor.fetchall()
for row in results20:
    print(row)
```

```python
def execute_query(query, params=None):
    if params is None:
        cursor.execute(query)
    else:
        cursor.execute(query, params)

    return cursor.fetchall()
```

```python
results = {}

for i in range(1, 21):
    query_name = f"query{i}"

    try:
        results[query_name] = execute_query(globals()[query_name])
        print(f"{query_name} executed successfully")

    except Exception as e:
        print(f"{query_name} failed")
        print(e)
        break
```
